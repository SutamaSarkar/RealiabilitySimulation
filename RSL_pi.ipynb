```python
import random
import matplotlib.pyplot as plt
import math

#trials graph
trial_list=[1000,10000,100000]
N_list= []
incircle_list= []
prob_list= []
pi_list= []


for N in trial_list:

    in_circle=0
    x_in= []
    y_in= []
    x_out= []
    y_out= []

    for i in range (N):
        x = random.uniform(-0.5, 0.5)
        y = random.uniform(-0.5, 0.5)

        distance = math.sqrt(x*x + y*y)

        if distance <= 0.5:
            in_circle += 1
            x_in.append(x)
            y_in.append(y)
        else:
            x_out.append(x)
            y_out.append(y)
    #print("in_circle values",in_circle)
    z = in_circle / N
    pi=4 * z
    #print("pi val is =",pi)

    N_list.append(N)
    incircle_list.append(in_circle)
    prob_list.append(z)
    pi_list.append(pi)

print("N=",N,"INCIRCLE =",in_circle,"PROB=",z,"PI=",pi)
print("\nTRIALS(N)\tINCIRCLE(in_circle)\tPROB(IN_C/N)\tPI(pi)")
for i in range(len(N_list)):
    print(f"{N_list[i]}\t\t\t{incircle_list[i]}\t\t\t\t{prob_list[i]:.6f}\t\t{pi_list[i]:.6f}")

plt.figure(figsize=(5,5))
plt.plot(N_list, pi_list, marker="o", label="Estimated Pi")
plt.axhline(y=math.pi, linestyle="--", label="Actual Pi")

plt.xscale("log")
plt.xlabel("Number of Trials (log scale)")
plt.ylabel("Pi Estimate")
plt.title("Monte Carlo Convergence of Pi")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import random
import matplotlib.pyplot as plt
import math

# ---------------- TRIAL VALUES ----------------
trial_list = [1000, 10000, 100000]

# ---------------- STORAGE LISTS ----------------
N_list = []
incircle_list = []
prob_list = []
pi_list = []

# ---------------- MONTE CARLO SIMULATION ----------------
for N in trial_list:

    in_circle = 0

    x_in, y_in = [], []
    x_out, y_out = [], []

    for _ in range(N):

        x = random.uniform(-0.5, 0.5)
        y = random.uniform(-0.5, 0.5)

        distance = math.sqrt(x**2 + y**2)

        if distance <= 0.5:
            in_circle += 1
            x_in.append(x)
            y_in.append(y)
        else:
            x_out.append(x)
            y_out.append(y)

    probability = in_circle / N
    pi_estimate = 4 * probability

    N_list.append(N)
    incircle_list.append(in_circle)
    prob_list.append(probability)
    pi_list.append(pi_estimate)

# ---------------- PRINT RESULTS ----------------
print("\nTRIALS(N)\tINCIRCLE\tPROBABILITY\tPI ESTIMATE")
print("-"*55)

for i in range(len(N_list)):
    print(f"{N_list[i]:<10}\t{incircle_list[i]:<8}\t{prob_list[i]:.6f}\t{pi_list[i]:.6f}")

# ---------------- PLOT CONVERGENCE ----------------
plt.figure(figsize=(5,5))

plt.plot(N_list, pi_list, marker="o", label="Estimated π")
plt.axhline(y=math.pi, linestyle="--", label="Actual π")

plt.xscale("log")
plt.xlabel("Number of Trials (log scale)")
plt.ylabel("π Estimate")
plt.title("Monte Carlo Convergence of π")
plt.legend()
plt.grid(True)

plt.show()
```
